```python
import requests

response = requests.get("https://api.open-meteo.com/v1/forecast?latitude=51.5074&longitude=-0.1278&current_weather=true")
print(response.status_code)
print(response.json())
```

```python
weather_of_london = response.json()["current_weather"]["temperature"]
print(weather_of_london)
```

```python
city_name = input("Which country's weather would you like to know today? ")
url = f"https://geocoding-api.open-meteo.com/v1/search?name={city_name}&count=10&language=en&format=json"
response2 = requests.get(url)
latitude = response2.json()["results"][0]["latitude"]
longitude = response2.json()["results"][0]["longitude"]
response = requests.get(f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current_weather=true")
print(f"The Temprature of {city_name} is ", response.json()["current_weather"]["temperature"])
print(f"The Windspeed of {city_name} is ",response.json()["current_weather"]["windspeed"])
```
