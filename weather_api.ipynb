```python
#Version 3
import numpy as np
city_name=input("Enter a city: ")
response3 = requests.get(f"https://geocoding-api.open-meteo.com/v1/search?name={city_name}&count=10&language=en&format=json")
lat = response3.json()["results"][0]["latitude"]
long = response3.json()["results"][0]["longitude"]
forcast = requests.get(f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={long}&daily=temperature_2m_max&current=temperature_2m&past_days=3&forecast_days=3")
C_temp = forcast.json()["current"]["temperature_2m"]
past = forcast.json()["daily"]["temperature_2m_max"][:3]
future = forcast.json()["daily"]["temperature_2m_max"][4:]
def avg_of_temps(past,future):
    past_avg = round(float(np.mean(past)))
    future_avg = round(float(np.mean(future)))
    return f'The temp for the past few days for around {past_avg} and the forcast temp is around {future_avg}'
print(f'Current temp is {C_temp}')
print(avg_of_temps(past,future))
```
